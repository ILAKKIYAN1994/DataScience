####Task 2 - Writing Analytical Queries


#####Revenue Trend (Daily)

```python
%sql
SELECT
  date(event_time) AS event_date,
  SUM(price) AS daily_revenue
FROM workspace.default.events_table
WHERE event_type = 'purchase'
GROUP BY event_date
ORDER BY event_date;
```

#####Funnel Analysis (View → Purchase)

```python
%sql
SELECT
  COUNT(DISTINCT CASE WHEN event_type = 'view' THEN user_id END) AS views,
  COUNT(DISTINCT CASE WHEN event_type = 'purchase' THEN user_id END) AS purchases,
  ROUND(
    COUNT(DISTINCT CASE WHEN event_type = 'purchase' THEN user_id END) * 100.0 /
    COUNT(DISTINCT CASE WHEN event_type = 'view' THEN user_id END),
    2
  ) AS conversion_rate
FROM workspace.default.events_table;
```

####Task 3 - Building Dashboards
- Run a SQL query
- Click Add visualization
- Choose chart type:
- Line → revenue trends
- Bar → top products
- KPI → funnel metrics
- Save the visualization
- Click Create dashboard
- Add multiple visualizations

#####Top Products by Revenue

```python
%sql
SELECT
  product_id,
  SUM(price) AS revenue
FROM workspace.default.events_table
WHERE event_type = 'purchase'
GROUP BY product_id
ORDER BY revenue DESC
LIMIT 10;
```

####Task 4 - Adding Filters
Filters are added at the dashboard level, not query level. But we also add at query level as shown below

```python
%sql
SELECT
  date(event_time) AS event_date,
  SUM(price) AS revenue
FROM workspace.default.events_table
WHERE event_type = 'purchase'
AND event_time BETWEEN '2019-10-01' AND '2019-10-31'
GROUP BY event_date
ORDER BY event_date;
```
