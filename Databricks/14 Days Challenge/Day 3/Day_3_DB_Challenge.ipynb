###We will be loading kaggle dataset loaded already into the volume

```python
events = spark.read.csv(\
    "/Volumes/workspace/ecommerce/ecommerce_data",
    header=True,
    inferSchema=True
)
display(events)
```

###Importing Window function

```python
from pyspark.sql import functions as F
from pyspark.sql.window import Window
```

###Top 5 products by revenue

```python
revenue = events.filter(F.col("event_type") == "purchase") \
    .groupBy("product_id") \
    .agg(F.round(F.sum("price"), 2).alias("revenue"))\
    .orderBy(F.desc("revenue")).limit(5)
 #Rounded to 2 decimal places for easier viewing\ 
display(revenue)
# Spark only shows what is mentioned in the groupby. It avoids other columns since it don't know how to aggregate them
```

#### Running total per user

```python
# Partition data by each user and count the number of events for each user
window = Window.partitionBy("user_id").orderBy("event_time")
# Add a column to the events DataFrame that shows the number of events for each user
running_total = events.withColumn("cumulative_events", F.count("*").over(window))
display(running_total)
```

###Conversion rate per category

```python
#The original code had sum(count) after pivot which is wrong. The correct code is below.
conversion_rate = (events.groupBy("category_code")\
                    .pivot("event_type").count()\
                    .withColumn("Conversion_rate",F.format_string("%.2f%%",(F.col("purchase")/F.col("view"))*100)))

display(conversion_rate)
#Formatted conversion rate to as percentage
```
