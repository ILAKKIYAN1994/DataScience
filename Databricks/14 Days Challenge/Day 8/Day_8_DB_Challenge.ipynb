#DAY 8 Databricks Challenge

####Task 1 - Creating Catalogue and Structures
#####Catalog -> Schema(Databases) -> Table/View
#####Purpose: Data Isolation, Fine-grained access control, enables lineage and auditing

```python
%sql
CREATE CATALOG IF NOT EXISTS ecommerce;
USE CATALOG ecommerce;

CREATE SCHEMA IF NOT EXISTS bronze;
CREATE SCHEMA IF NOT EXISTS silver;
CREATE SCHEMA IF NOT EXISTS gold;
```

#####Verifying the created schemas

```python
%sql
USE CATALOG ecommerce;
SHOW SCHEMAS;
```

#**************************************************

####Task 2 - Register Delta Tables 
#####Delta tables are type of external tables which means when the table is deleted, only metadata is deleted and not the actual data

```python
%sql
CREATE TABLE bronze.events
USING DELTA
AS SELECT * FROM delta.`/Volumes/workspace/default/challenge/bronze_events`;

CREATE TABLE silver.events
USING DELTA
AS SELECT * FROM delta.`/Volumes/workspace/default/challenge/silver_events`;

CREATE TABLE gold.products
USING DELTA
AS SELECT * FROM delta.`/Volumes/workspace/default/challenge/gold_products`;

--Community Edition do not support Unity Catalog.
```
