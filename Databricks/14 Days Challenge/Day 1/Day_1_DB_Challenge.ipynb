#Basic PySpark Commands

###Check Spark is Running

```python
spark
```

###Simple DataFrame


```python
data = [
    ("iPhone", "Electronics", 999),
    ("Samsung", "Electronics", 799),
    ("MacBook", "Electronics", 1299),
    ("AirPods", "Accessories", 199)
]

df = spark.createDataFrame(data, ["product", "category", "price"])
df.show()
```

###View Schema

```python
df.printSchema()
```

###Select Specific Columns

```python
df.select("product").show()
```

###Filtering based on prices

```python
df.filter(df.price > 1000).show()

#Equivalent Syntax

df.filter("price > 500").show()
```

###Adding New Column

```python
from pyspark.sql.functions import col
# WithColumn creates a new column with the name and expression
df.withColumn("price_with_tax", col("price") * 1.18).show()
```

###Rename a Column

```python
df.withColumnRenamed("price", "price_INR").show()
```

###Sorting Column

```python
#OrderBy sorts the dataframe by the column name
df.orderBy("price", ascending=False).show()
#Default sorting is ascending
```

###Counting Rows

```python
df.select("category").distinct().show()
```

###Group and Aggregate

```python
from pyspark.sql.functions import avg, max

df.groupBy("category").agg(
    avg("price").alias("avg_price"),
    max("price").alias("max_price")
).show()
```

###Check for Null values

```python
df.filter(df.product.isNull()).show()
```

###Show limited rows as per our need

```python
df.show(5, truncate=False)
```
