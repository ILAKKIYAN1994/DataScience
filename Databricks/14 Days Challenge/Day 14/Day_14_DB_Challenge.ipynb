####Task 1 - Use Genie to query data with natural language
#####Databricks Genie enables natural language analytics by translating business questions into optimized SQL queries executed directly on the Lakehouse.

#####Query 1.1 - Show me total revenue by product id

```python
%sql
SELECT
  product_id,
  SUM(revenue) AS total_revenue
FROM
  workspace.gold.products
GROUP BY
  product_id
ORDER BY
  product_id
```

#####Query 1.2 - Show me total revenue by product id

```python
%sql
SELECT
  product_id,
  SUM(conversion_rate) / NULLIF(SUM(views), 0) AS conversion_rate
FROM
  workspace.gold.products
GROUP BY
  product_id
ORDER BY
  conversion_rate DESC
```

####Task 3 - Build simple NLP task

We’ll practice AI-powered analysis using:

- HuggingFace
- Transformers
- MLflow tracking

This mirrors Mosaic AI workflows conceptually as Databricks CE don't support Mosaic AI

#####Step 3.1 - Installing Transformers and Torch packages

```python
%pip install transformers torch

```

```python
#Loading pre-trained model
from transformers import pipeline

classifier = pipeline("sentiment-analysis")
```

#####Step 3.2 - Analyze the text data

```python
reviews = [
    "This product is amazing!",
    "Terrible quality, waste of money",
    "Good value for the price",
    "Not worth buying"
]

results = classifier(reviews)
results
```

#####Step 3.3 - Convert Results into Insights

```python
for review, result in zip(reviews, results):
    print(f"Review: {review}")
    print(f"Sentiment: {result['label']} (score={result['score']:.2f})")
    print("-"*50)
```

#####Step 3.4 - Log AI Experiment to MLflow

```python
#Logging the model into MLflow
import mlflow

with mlflow.start_run(run_name="sentiment_analysis_day14"):
    mlflow.log_param("model", "distilbert-base-uncased-finetuned-sst-2")
    mlflow.log_param("task", "sentiment-analysis")

    # Example metric
    mlflow.log_metric("sample_accuracy", 0.95)
```
